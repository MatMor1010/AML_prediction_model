### Which column type to transform?

We will need all the binary indicator as categorical instead of float or integer for the Target (rating)

```python
# Working with modifier df -- drop ID col
df_mod = df.drop(columns = "cust_id_masked")

# Binary columns identifier
binary_cols = df_mod.columns[df_mod.nunique() == 2]
list(binary_cols)

# All of those will be put categorical
```

```python
# Its a freaking pain, binary but not already 0 - 1
df_mod["cust_cdn_resident"]
```

```python
# Lets put them 0 - 1; then categorical
for col in binary_cols:
    df_mod[col] = (df_mod[col] == df_mod[col].max()).astype(int)
    
    df_mod[col] = df_mod[col].astype("category")
```

Lets check if any other columns are exactly the same value of another column in the dataset:

```python
similar_col = {}
cols = df_mod.columns

for i, col1 in enumerate(cols):
    for col2 in cols[i + 1:]:
        if df_mod[col1].equals(df_mod[col2]):
            similar_col.setdefault(col1, []).append(col2)

similar_col
```
